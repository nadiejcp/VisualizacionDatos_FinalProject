```python
import pandas as pd
import altair as alt
import os
import shutil
```

```python
df = pd.read_csv('data/raw/data.csv')
df
```

```python
df['category'].unique()
```

```python
processed_values = {
  'carbon_dioxide_co2_emissions_without_land_use_land_use_change_and_forestry_lulucf_in_kilotonne_co2_equivalent': 'CO2 Emissions',
  'greenhouse_gas_ghgs_emissions_including_indirect_co2_without_lulucf_in_kilotonne_co2_equivalent': 'GHG Emissions with CO2',
  'greenhouse_gas_ghgs_emissions_without_land_use_land_use_change_and_forestry_lulucf_in_kilotonne_co2_equivalent': 'GHG Emissions',
  'hydrofluorocarbons_hfcs_emissions_in_kilotonne_co2_equivalent': 'HFC Emissions',
  'methane_ch4_emissions_without_land_use_land_use_change_and_forestry_lulucf_in_kilotonne_co2_equivalent': 'CH4 Emissions',
  'nitrogen_trifluoride_nf3_emissions_in_kilotonne_co2_equivalent': 'NF3 Emissions',
  'nitrous_oxide_n2o_emissions_without_land_use_land_use_change_and_forestry_lulucf_in_kilotonne_co2_equivalent': 'N2O Emissions',
  'perfluorocarbons_pfcs_emissions_in_kilotonne_co2_equivalent': 'PFC Emissions',
  'sulphur_hexafluoride_sf6_emissions_in_kilotonne_co2_equivalent': 'SF6 Emissions',
  'unspecified_mix_of_hydrofluorocarbons_hfcs_and_perfluorocarbons_pfcs_emissions_in_kilotonne_co2_equivalent': 'Unspecified Emissions'
}
df['emission'] = df['category'].map(processed_values)
df['units'] = 'Kilotonne CO2 Equivalent'
df = df.drop(columns=['category'])
df = df.rename(columns={'country_or_area': 'country'})
```

```python
df.to_csv('data/processed/data.csv', index=False)
```

```python
df_total = df.groupby(['emission', 'year'])['value'].sum().reset_index()
df_total['units'] = 'Kilotonne CO2 Equivalent'
```

```python
df_total
```

```python
alt.Chart(df_total).mark_line().encode(
  x=alt.X("year:Q", bin=False, title="Year"),
  y=alt.Y("value:Q", title="Kilotonne CO2 Equivalent")
).facet(
  column=alt.Column("emission:N", title="Emission Type")
)
```

```python
df_total.to_csv('data/processed/total_emissions.csv', index=False)
```

```python
df_country = df.groupby(['country', 'year'])['value'].sum().reset_index()
df_country['units'] = 'Kilotonne CO2 Equivalent'
df_country = df_country[df_country['country'] != 'European Union']
```

```python
df_country
```

```python
alt.Chart(df_country).mark_line().encode(
  x=alt.X("year:Q", bin=False, title="Year"),
  y=alt.Y("value:Q", title="Kilotonne CO2 Equivalent")
).facet(
  column=alt.Column("country:N", title="Country")
)
```

```python
df_country.to_csv('data/processed/country_emissions.csv', index=False)
```
